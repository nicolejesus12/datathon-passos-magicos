```python
mapa_fase_2023 = {
    "ALFA": 0,
    "FASE 1": 1,
    "FASE 2": 2,
    "FASE 3": 3,
    "FASE 4": 4,
    "FASE 5": 5,
    "FASE 6": 6,
    "FASE 7": 7,
    "FASE 8": 8
}

df2023["fase"] = df2023["fase"].replace(mapa_fase_2023)
```

```python
def converter_fase(valor):
    if pd.isna(valor):
        return None

    valor = str(valor).upper().strip()

    if valor == "ALFA":
        return 0

    if valor[0].isdigit():
        return int(valor[0])

    return None

df2024["fase"] = df2024["fase"].apply(converter_fase)
```

Entretanto, como a correlação não estabelece uma relação de causa e efeito, os resultados devem ser interpretados como uma evidência de associação entre as variáveis, indicando que o engajamento pode ser um fator relevante para o acompanhamento do desempenho dos estudantes.

Assim, alunos que demonstram elevada confiança em sua percepção pessoal não necessariamente obtêm melhores resultados acadêmicos, da mesma forma que estudantes com desempenho elevado podem apresentar autoavaliações mais modestas.
